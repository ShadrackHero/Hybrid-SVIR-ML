```python
# Adjusting the path if needed so it points to raw file
df = pd.read_csv("../data/raw/owid_vaccinations_south_africa.csv")
cases = pd.read_csv("../data/raw/covid19za_provincial_confirmed.csv")
print("Shape of full dataset:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())



print("Shape:", cases.shape)
print("\nColumns:")
print(cases.columns.tolist())
print("\nFirst 5 rows:")
print(cases.head())
```

```python
# Look at date range and a few more details
print("Date range (raw):", cases["date"].min(), "→", cases["date"].max())
print("\nMissing values per column:")
print(cases.isna().sum())

print("\nLast 5 rows:")
print(cases.tail())
```

```python
# Parse the date column (DD-MM-YYYY format)
cases["date"] = pd.to_datetime(cases["date"], format="%d-%m-%Y")

print("True date range:", cases["date"].min().date(), "→", cases["date"].max().date())
print("Number of days:", (cases["date"].max() - cases["date"].min()).days + 1)
print("Number of rows:", len(cases))

# Quick check that total is cumulative
print("\nFirst few totals:", cases["total"].head(3).tolist())
print("Last few totals:", cases["total"].tail(3).tolist())
```

```python
# Create a continuous daily date range
full_dates = pd.date_range(start=cases["date"].min(), 
                           end=cases["date"].max(), 
                           freq="D")

# Reindex so every day is present
cases = cases.set_index("date").reindex(full_dates)
cases.index.name = "date"
cases = cases.reset_index()

# Forward-fill the cumulative total (safe for cumulative series)
cases["total"] = cases["total"].ffill()

# Compute daily new cases
cases["daily_cases"] = cases["total"].diff().fillna(0)

print("Shape after reindexing:", cases.shape)
print("Missing days filled:", cases["total"].isna().sum())  # should be 0 after ffill
print("\nFirst 10 rows of daily_cases:")
print(cases[["date", "total", "daily_cases"]].head(10))
print("\nSome later rows:")
print(cases[["date", "total", "daily_cases"]].iloc[100:105])
```

```python
# Load vaccination data (if not already loaded)
vax = pd.read_csv("../data/raw/owid_vaccinations_south_africa.csv")

# Parse dates
vax["date"] = pd.to_datetime(vax["date"])

print("Vaccination date range:", vax["date"].min().date(), "→", vax["date"].max().date())
print("Shape:", vax.shape)
print("\nMissing values in key vaccination columns:")
print(vax[["people_vaccinated", "people_fully_vaccinated", "total_vaccinations", 
           "daily_vaccinations"]].isna().sum())

print("\nFirst few rows:")
print(vax[["date", "people_vaccinated", "people_fully_vaccinated", "daily_vaccinations"]].head())
print("\nLast few rows:")
print(vax[["date", "people_vaccinated", "people_fully_vaccinated", "daily_vaccinations"]].tail())
```

```python
# Forward-fill cumulative vaccination columns
vax_cols = ["people_vaccinated", "people_fully_vaccinated", "total_vaccinations"]
for c in vax_cols:
    vax[c] = vax[c].ffill()

# Set any remaining early NaNs to 0 (before the programme started)
vax[vax_cols] = vax[vax_cols].fillna(0)

# Also fill the single missing daily_vaccinations if present
vax["daily_vaccinations"] = vax["daily_vaccinations"].fillna(0)

print("Missing values after cleaning:")
print(vax[vax_cols + ["daily_vaccinations"]].isna().sum())

print("\nFirst few rows after cleaning:")
print(vax[["date", "people_vaccinated", "people_fully_vaccinated", "daily_vaccinations"]].head(6))
print("\nLast few rows:")
print(vax[["date", "people_vaccinated", "people_fully_vaccinated", "daily_vaccinations"]].tail(3))
```

```python
# Merging on date (inner join keeps only the overlapping period)
merged = pd.merge(
    cases[["date", "total", "daily_cases"]],
    vax[["date", "people_vaccinated", "people_fully_vaccinated", 
         "total_vaccinations", "daily_vaccinations"]],
    on="date",
    how="inner"
)

print("Merged shape:", merged.shape)
print("Merged date range:", merged["date"].min().date(), "→", merged["date"].max().date())

print("\nFirst few rows:")
print(merged.head())
print("\nLast few rows:")
print(merged.tail())
print("\nMissing values in merged table:")
print(merged.isna().sum())
```

```python
print("Peak daily cases:", merged["daily_cases"].max())
print("Date of peak daily cases:", merged.loc[merged["daily_cases"].idxmax(), "date"].date())

print("\nFinal cumulative cases:", merged["total"].iloc[-1])
print("Final people vaccinated:", merged["people_vaccinated"].iloc[-1])
print("Final people fully vaccinated:", merged["people_fully_vaccinated"].iloc[-1])

# Rough vaccination coverage (using approximate SA population ~60 million)
pop = 60_000_000
print(f"\nApprox. full vaccination coverage at end: {merged['people_fully_vaccinated'].iloc[-1] / pop:.1%}")

# Save the cleaned merged dataset
merged.to_csv("../data/processed/south_africa_cases_vaccination_cleaned.csv", index=False)
print("\nSaved to data/processed/south_africa_cases_vaccination_cleaned.csv")
```
