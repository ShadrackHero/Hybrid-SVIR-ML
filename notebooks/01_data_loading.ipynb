```python
from pathlib import Path

import pandas as pd

def find_project_root(marker="data"):
    current = Path.cwd().resolve()
    for folder in [current, *current.parents]:
        if (folder / marker).exists():
            return folder
    raise FileNotFoundError(f"No parent folder of {current} contains a '{marker}' directory")

RAW_DIR = find_project_root() / "data" / "raw"

datasets = {
    "confirmed":        "covid19za_provincial_confirmed.csv",
    "deaths":           "covid19za_provincial_deaths.csv",
    "vaccination":      "covid19za_provincial_vaccination.csv",
    "owid_vaccination": "owid_vaccinations_south_africa.csv",
}

confirmed, deaths, vaccination, owid_vaccination = (
    pd.read_csv(RAW_DIR / datasets[k])
    for k in ("confirmed", "deaths", "vaccination", "owid_vaccination")
)

for name, df in zip(datasets, (confirmed, deaths, vaccination, owid_vaccination)):
    print(f"{name:<18} shape={df.shape}")

print("\nConfirmed columns:", confirmed.columns.tolist())
```
